```python
import cv2
import numpy as np

sharpness = []

for path in images:
    img = cv2.imread(str(path), cv2.IMREAD_GRAYSCALE)
    if img is None:
        continue

    score = cv2.Laplacian(img, cv2.CV_64F).var()
    sharpness.append((path.name, float(score)))

values = np.array([s for _, s in sharpness])

print("=" * 70)
print("IMAGE SHARPNESS CHECK")
print("=" * 70)
print("Images measured :", len(values))
print("Minimum         :", round(values.min(), 2))
print("Median          :", round(np.median(values), 2))
print("Mean            :", round(values.mean(), 2))
print("Maximum         :", round(values.max(), 2))

for percentile in [5, 10, 25, 75, 90, 95]:
    print(f"P{percentile:<2}            :", round(np.percentile(values, percentile), 2))

print("\nSharpest 5:")
for name, score in sorted(sharpness, key=lambda x: x[1], reverse=True)[:5]:
    print(f"  {name}: {score:.2f}")

print("\nSoftest 5:")
for name, score in sorted(sharpness, key=lambda x: x[1])[:5]:
    print(f"  {name}: {score:.2f}")
```

```python
import h5py
import cv2
import numpy as np
import json

# Original capture metadata on Google Drive
TRANSFORMS = DATASET / "transforms.json"

with open(TRANSFORMS, "r") as f:
    transforms = json.load(f)

fx = transforms["fl_x"]
fy = transforms["fl_y"]
cx = transforms["cx"]
cy = transforms["cy"]

K = np.array([
    [fx, 0, cx],
    [0, fy, cy],
    [0,  0, 1]
], dtype=np.float64)

print("=" * 70)
print("SUPERGLUE GEOMETRIC VERIFICATION")
print("=" * 70)
print("Intrinsics:")
print(K)

verified_pairs = 0
total_raw = 0
total_inliers = 0
inlier_counts = []

with h5py.File(feature_path, "r") as feat_file, \
     h5py.File(MATCH_FILE, "r") as match_file:

    for pair in open(PAIR_FILE):
        name0, name1 = pair.strip().split()

        if name0 not in match_file:
            continue

        if name1 not in match_file[name0]:
            continue

        matches0 = match_file[name0][name1]["matches0"][:]

        idx0 = np.where(matches0 >= 0)[0]

        if len(idx0) < 8:
            continue

        idx1 = matches0[idx0].astype(np.int32)

        pts0 = feat_file[name0]["keypoints"][:][idx0].astype(np.float64)
        pts1 = feat_file[name1]["keypoints"][:][idx1].astype(np.float64)

        E, mask = cv2.findEssentialMat(
            pts0,
            pts1,
            K,
            method=cv2.RANSAC,
            prob=0.999,
            threshold=1.5
        )

        if E is None or mask is None:
            continue

        mask = mask.ravel().astype(bool)
        inliers = int(mask.sum())

        total_raw += len(idx0)
        total_inliers += inliers
        inlier_counts.append(inliers)

        if inliers >= 15:
            verified_pairs += 1

print()
print("Pairs with ≥15 inliers :", verified_pairs)
print("Raw correspondences    :", total_raw)
print("Geometric inliers      :", total_inliers)

if total_raw:
    print("Overall inlier ratio    :", round(total_inliers / total_raw, 3))

if inlier_counts:
    print("Median inliers/pair     :", int(np.median(inlier_counts)))
    print("Maximum inliers/pair    :", max(inlier_counts))

print()
print("✅ Geometric verification completed")
```

```python
# CELL 23 — HLoc/COLMAP → Nerfstudio transforms.json

import json
import numpy as np
from pathlib import Path

NS_DATA = WORK / "nerfstudio_data"
NS_DATA.mkdir(parents=True, exist_ok=True)

# ------------------------------------------------------------------
# COLMAP camera model → Nerfstudio camera model
# ------------------------------------------------------------------

def convert_camera(cam):
    model = cam.model.name
    p = np.asarray(cam.params, dtype=float)

    if model == "SIMPLE_RADIAL":
        # COLMAP: [f, cx, cy, k]
        f, cx, cy, k = p

        return {
            "camera_model": "OPENCV",
            "fl_x": float(f),
            "fl_y": float(f),
            "cx": float(cx),
            "cy": float(cy),
            "k1": float(k),
            "k2": 0.0,
            "p1": 0.0,
            "p2": 0.0,
        }

    elif model == "PINHOLE":
        # COLMAP: [fx, fy, cx, cy]
        fx, fy, cx, cy = p

        return {
            "camera_model": "OPENCV",
            "fl_x": float(fx),
            "fl_y": float(fy),
            "cx": float(cx),
            "cy": float(cy),
            "k1": 0.0,
            "k2": 0.0,
            "p1": 0.0,
            "p2": 0.0,
        }

    else:
        raise ValueError(f"Unsupported camera model: {model}")


# ------------------------------------------------------------------
# Build frames
# ------------------------------------------------------------------

frames = []

for img in registered:
    cam = sfm_model.cameras[img.camera_id]

    # COLMAP / HLoc pose:
    # x_camera = R * x_world + t
    T_cw = np.asarray(img.cam_from_world().matrix(), dtype=float)

    # Nerfstudio expects camera-to-world
    T_wc = np.linalg.inv(T_cw)

    camera_info = convert_camera(cam)

    frame = {
        "file_path": f"images/{img.name}",
        "transform_matrix": T_wc.tolist(),
        **camera_info,
        "w": int(cam.width),
        "h": int(cam.height),
    }

    frames.append(frame)


# ------------------------------------------------------------------
# Dataset metadata
# ------------------------------------------------------------------

# Use the first camera as the global camera model reference.
# Per-frame intrinsics are retained in each frame.
first_cam = sfm_model.cameras[registered[0].camera_id]
first_info = convert_camera(first_cam)

transforms = {
    "camera_model": first_info["camera_model"],
    "fl_x": first_info["fl_x"],
    "fl_y": first_info["fl_y"],
    "cx": first_info["cx"],
    "cy": first_info["cy"],
    "k1": first_info["k1"],
    "k2": first_info["k2"],
    "p1": first_info["p1"],
    "p2": first_info["p2"],
    "w": int(first_cam.width),
    "h": int(first_cam.height),
    "frames": frames,
}

OUT_JSON = NS_DATA / "transforms.json"

with open(OUT_JSON, "w") as f:
    json.dump(transforms, f, indent=2)

print("=" * 70)
print("NERFSTUDIO DATASET CREATED")
print("=" * 70)

print(f"Output directory : {NS_DATA}")
print(f"transforms.json  : {OUT_JSON}")
print(f"Registered frames: {len(frames)}")
print(f"Image resolution : {first_cam.width} x {first_cam.height}")
print(f"Camera model     : {first_info['camera_model']}")
print(f"First focal      : {first_info['fl_x']:.2f}")
print(f"First principal  : ({first_info['cx']:.2f}, {first_info['cy']:.2f})")

print("\nFirst frame:")
print(frames[0]["file_path"])
print("Pose:")
print(np.array(frames[0]["transform_matrix"]))

print("\n✓ HLoc → Nerfstudio conversion complete")
```

```python
# CELL 23 — HLoc/COLMAP → Nerfstudio transforms.json

import json
import numpy as np
from pathlib import Path

NS_DATA = WORK / "nerfstudio_data"
NS_DATA.mkdir(parents=True, exist_ok=True)


def convert_camera(cam):
    model = cam.model.name
    p = np.asarray(cam.params, dtype=float)

    if model == "SIMPLE_RADIAL":
        # COLMAP: [f, cx, cy, k]
        f, cx, cy, k = p

        return {
            "camera_model": "OPENCV",
            "fl_x": float(f),
            "fl_y": float(f),
            "cx": float(cx),
            "cy": float(cy),
            "k1": float(k),
            "k2": 0.0,
            "p1": 0.0,
            "p2": 0.0,
        }

    elif model == "PINHOLE":
        fx, fy, cx, cy = p

        return {
            "camera_model": "OPENCV",
            "fl_x": float(fx),
            "fl_y": float(fy),
            "cx": float(cx),
            "cy": float(cy),
            "k1": 0.0,
            "k2": 0.0,
            "p1": 0.0,
            "p2": 0.0,
        }

    else:
        raise ValueError(f"Unsupported camera model: {model}")


frames = []

for img in registered:
    cam = sfm_model.cameras[img.camera_id]

    # COLMAP camera-from-world transform
    pose = img.cam_from_world()

    # Rigid3d rotation and translation
    R_cw = np.asarray(pose.rotation.matrix(), dtype=float)
    t_cw = np.asarray(pose.translation, dtype=float).reshape(3)

    # Build homogeneous world -> camera matrix
    T_cw = np.eye(4, dtype=float)
    T_cw[:3, :3] = R_cw
    T_cw[:3, 3] = t_cw

    # Nerfstudio needs camera -> world
    T_wc = np.linalg.inv(T_cw)

    camera_info = convert_camera(cam)

    frames.append({
        "file_path": f"images/{img.name}",
        "transform_matrix": T_wc.tolist(),
        **camera_info,
        "w": int(cam.width),
        "h": int(cam.height),
    })


# Use first camera as dataset-level defaults
first_cam = sfm_model.cameras[registered[0].camera_id]
first_info = convert_camera(first_cam)

transforms = {
    "camera_model": first_info["camera_model"],
    "fl_x": first_info["fl_x"],
    "fl_y": first_info["fl_y"],
    "cx": first_info["cx"],
    "cy": first_info["cy"],
    "k1": first_info["k1"],
    "k2": first_info["k2"],
    "p1": first_info["p1"],
    "p2": first_info["p2"],
    "w": int(first_cam.width),
    "h": int(first_cam.height),
    "frames": frames,
}

OUT_JSON = NS_DATA / "transforms.json"

with open(OUT_JSON, "w") as f:
    json.dump(transforms, f, indent=2)

print("=" * 70)
print("NERFSTUDIO DATASET CREATED")
print("=" * 70)

print(f"Output directory : {NS_DATA}")
print(f"transforms.json  : {OUT_JSON}")
print(f"Registered frames: {len(frames)}")
print(f"Image resolution : {first_cam.width} x {first_cam.height}")
print(f"Camera model     : {first_info['camera_model']}")
print(f"First focal      : {first_info['fl_x']:.2f}")
print(f"Principal point  : ({first_info['cx']:.2f}, {first_info['cy']:.2f})")

print("\nFirst frame:", frames[0]["file_path"])
print("Camera-to-world pose:")
print(np.array(frames[0]["transform_matrix"]))

print("\n✓ HLoc → Nerfstudio conversion complete")
```

```python
# CELL 24 — Validate Nerfstudio dataset

import json
import numpy as np
from pathlib import Path

with open(NS_DATA / "transforms.json", "r") as f:
    ns = json.load(f)

print("=" * 70)
print("NERFSTUDIO DATASET VALIDATION")
print("=" * 70)

frames = ns["frames"]

missing = []
bad_pose = []
bad_intrinsics = []

for frame in frames:
    # Check image
    image_path = NS_DATA / frame["file_path"]
    if not image_path.exists():
        missing.append(frame["file_path"])

    # Check pose
    T = np.asarray(frame["transform_matrix"], dtype=float)

    if T.shape != (4, 4) or not np.all(np.isfinite(T)):
        bad_pose.append(frame["file_path"])
    elif not np.allclose(T[3], [0, 0, 0, 1], atol=1e-5):
        bad_pose.append(frame["file_path"])

    # Check intrinsics
    for key in ["fl_x", "fl_y", "cx", "cy"]:
        if key not in frame or not np.isfinite(frame[key]):
            bad_intrinsics.append(frame["file_path"])
            break

print(f"Frames                 : {len(frames)}")
print(f"Missing images         : {len(missing)}")
print(f"Bad pose matrices      : {len(bad_pose)}")
print(f"Bad intrinsics         : {len(bad_intrinsics)}")

if missing:
    print("\nMissing:")
    print(missing[:10])

if bad_pose:
    print("\nBad poses:")
    print(bad_pose[:10])

if bad_intrinsics:
    print("\nBad intrinsics:")
    print(bad_intrinsics[:10])

# Basic camera-center extent
centers = []

for frame in frames:
    T = np.asarray(frame["transform_matrix"], dtype=float)
    centers.append(T[:3, 3])

centers = np.asarray(centers)

print("\nCamera trajectory bounds:")
print(f"X: {centers[:,0].min():.3f} to {centers[:,0].max():.3f}")
print(f"Y: {centers[:,1].min():.3f} to {centers[:,1].max():.3f}")
print(f"Z: {centers[:,2].min():.3f} to {centers[:,2].max():.3f}")

print("\n✓ Dataset validation complete")
```

```python
# CELL 13 — Convert the 119-image HLoc reconstruction to Nerfstudio format

import sys
import json
import numpy as np
from pathlib import Path

sys.path.insert(0, "/content/Hierarchical-Localization")

import pycolmap

WORK = Path("/content/digital_twin_3dgs")
RAW_IMAGES = WORK / "images"
SFM_OUTPUT = WORK / "hloc" / "sfm_full"
NS_DATA = WORK / "nerfstudio_data"

NS_DATA.mkdir(parents=True, exist_ok=True)

# Load the largest reconstruction
sfm = pycolmap.Reconstruction(str(SFM_OUTPUT))

print("Registered images:", len(sfm.images))
print("3D points:", len(sfm.points3D))

# Build Nerfstudio transforms
frames = []

registered = sorted(sfm.images.values(), key=lambda img: img.name)

for img in registered:
    cam = sfm.cameras[img.camera_id]

    # COLMAP camera parameters
    params = np.asarray(cam.params, dtype=float)

    if cam.model.name == "SIMPLE_RADIAL":
        f, cx, cy, k1 = params
        fl_x = fl_y = float(f)
        k2 = p1 = p2 = 0.0
    elif cam.model.name == "PINHOLE":
        fl_x, fl_y, cx, cy = params
        k1 = k2 = p1 = p2 = 0.0
    else:
        raise ValueError(f"Unsupported camera model: {cam.model.name}")

    # COLMAP: camera-from-world
    pose = img.cam_from_world()

    T_cw = np.eye(4)
    T_cw[:3, :3] = np.asarray(pose.rotation.matrix(), dtype=float)
    T_cw[:3, 3] = np.asarray(pose.translation, dtype=float).reshape(3)

    # Nerfstudio expects camera-to-world
    T_wc = np.linalg.inv(T_cw)

    frames.append({
        "file_path": f"images/{img.name}",
        "transform_matrix": T_wc.tolist(),
        "fl_x": float(fl_x),
        "fl_y": float(fl_y),
        "cx": float(cx),
        "cy": float(cy),
        "k1": float(k1),
        "k2": float(k2),
        "p1": float(p1),
        "p2": float(p2),
    })

transforms = {
    "camera_model": "OPENCV",
    "fl_x": frames[0]["fl_x"],
    "fl_y": frames[0]["fl_y"],
    "cx": frames[0]["cx"],
    "cy": frames[0]["cy"],
    "w": int(sfm.cameras[registered[0].camera_id].width),
    "h": int(sfm.cameras[registered[0].camera_id].height),
    "frames": frames,
}

with open(NS_DATA / "transforms.json", "w") as f:
    json.dump(transforms, f, indent=2)

# Link only the registered images
NS_IMAGES = NS_DATA / "images"
NS_IMAGES.mkdir(exist_ok=True)

for img in registered:
    src = RAW_IMAGES / img.name
    dst = NS_IMAGES / img.name
    if not dst.exists():
        dst.symlink_to(src)

print("\nNerfstudio dataset created.")
print("Transforms:", NS_DATA / "transforms.json")
print("Images:", len(list(NS_IMAGES.glob("*.jpg"))))
print("Resolution:", transforms["w"], "x", transforms["h"])
print("Camera model:", transforms["camera_model"])
```

```python
# CELL 14 — Fix Nerfstudio image directory to exactly match the 119-image SfM model

import shutil
from pathlib import Path

WORK = Path("/content/digital_twin_3dgs")
RAW_IMAGES = WORK / "images"
NS_DATA = WORK / "nerfstudio_data"
NS_IMAGES = NS_DATA / "images"
SFM_OUTPUT = WORK / "hloc" / "sfm_full"

import pycolmap

sfm = pycolmap.Reconstruction(str(SFM_OUTPUT))
registered = sorted(sfm.images.values(), key=lambda img: img.name)

# Remove stale image links from the previous 142-image reconstruction
if NS_IMAGES.exists():
    shutil.rmtree(NS_IMAGES)

NS_IMAGES.mkdir(parents=True)

# Create links only for images registered in the current SfM model
for img in registered:
    src = RAW_IMAGES / img.name
    dst = NS_IMAGES / img.name
    dst.symlink_to(src)

actual = sorted(p.name for p in NS_IMAGES.glob("*.jpg"))
expected = sorted(img.name for img in registered)

print("SfM registered:", len(expected))
print("Nerfstudio images:", len(actual))
print("Missing:", sorted(set(expected) - set(actual)))
print("Extra:", sorted(set(actual) - set(expected)))

assert actual == expected

print("\n✓ Nerfstudio dataset now exactly matches the 119-image SfM model.")
```
